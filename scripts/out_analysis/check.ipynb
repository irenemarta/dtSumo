```python
import sumolib
import sys
from pathlib import Path

ROOT = Path.cwd().parent.parent
if str(ROOT) not in sys.path:
    sys.path.append(str(ROOT))

from scripts.src.inputs import config as cfg

net = sumolib.net.readNet(cfg.NET_FILE)
max_l = 0
over_100_count = 0
over_100 = []
sub_15_count = 0
sub_15 = []

for edge in net.getEdges():
    l = edge.getLength()
    if l > max_l:
        max_l = l
    if l >= 100:
        over_100_count += 1 
        over_100.append({edge.getID(): l})
    elif l <= 15:
        sub_15_count += 1 
        sub_15.append({edge.getID(): l})
        
max_l, over_100_count, over_100
# sub_15_count
```
